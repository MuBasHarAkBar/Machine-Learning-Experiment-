```python

import numpy as np


import matplotlib.pyplot as plt


print("NumPy version:", np.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)

print("\nLibraries imported successfully!")
```

```python
rng = np.random.default_rng(7)
print("random generator inti. successfully")
```

```python
n = 80
d = 40
X = rng.normal(
    loc=0,
    scale=1,
    size=(n, d)
)
print("Input Matrix generated successfully")
print("Shape of X:", X.shape)
```

```python
print("First five Observation")
print(X[:5])
```

```python

epsilon = rng.normal(
    loc=0,
    scale=2,
    size=n
)

print("Noise generated successfully!")
print("Noise shape:", epsilon.shape)
```

```python
y = (
    1
    + 3 * X[:, 0]
    - 2 * X[:, 1]
    + 1.5 * X[:, 2]
    + 0.5 * X[:, 3]
    - X[:, 4]
    + epsilon
)
print("Target vector generated successfully!")
print("Target shape:", y.shape)
```

```python
print("First five target values:")
print(y[:5])
```

```python
indices = rng.permutation(n)

print("Shuffled indices:")
print(indices)
```

```python
train_indices = indices[:56]

test_indices = indices[56:]
```

```python
X_train = X[train_indices]
y_train = y[train_indices]

X_test = X[test_indices]
y_test = y[test_indices]

print("Dataset splitting completed!")
```

```python
print("Original dataset:")
print("X:", X.shape)
print("y:", y.shape)

print("\nTraining dataset:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting dataset:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
def calculate_mse(y_true, y_pred):

    mse = np.mean((y_true - y_pred) ** 2)

    return mse
```

```python
actual = np.array([10, 20, 30])
predicted = np.array([12, 18, 29])

result = calculate_mse(actual, predicted)

print("MSE:", result)
```

```python
X_train_5 = X_train[:, :5]

X_test_5 = X_test[:, :5]

print("Training features:", X_train_5.shape)
print("Testing features:", X_test_5.shape)
```

```python
X_train_5_bias = np.column_stack((
    np.ones(X_train_5.shape[0]),
    X_train_5
))

X_test_5_bias = np.column_stack((
    np.ones(X_test_5.shape[0]),
    X_test_5
))

print("Training matrix:", X_train_5_bias.shape)
print("Testing matrix:", X_test_5_bias.shape)
```

```python
A = X_train_5_bias.T @ X_train_5_bias

B = X_train_5_bias.T @ y_train

theta_5 = np.linalg.solve(A, B)

print("Estimated parameters:")
print(theta_5)
```

```python
print("Bias:", theta_5[0])

print("Weights:", theta_5[1:])
```

```python
y_train_pred_5 = X_train_5_bias @ theta_5

y_test_pred_5 = X_test_5_bias @ theta_5

print("Training predictions:")
print(y_train_pred_5[:5])

print("\nTesting predictions:")
print(y_test_pred_5[:5])
```

```python
train_mse_5 = calculate_mse(
    y_train,
    y_train_pred_5
)

test_mse_5 = calculate_mse(
    y_test,
    y_test_pred_5
)

print("Training MSE:", train_mse_5)
print("Testing MSE:", test_mse_5)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    y_test,
    y_test_pred_5,
    label="Predictions"
)

minimum = min(y_test.min(), y_test_pred_5.min())
maximum = max(y_test.max(), y_test_pred_5.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    'r--',
    label="Perfect prediction"
)

plt.xlabel("Actual values")
plt.ylabel("Predicted values")

plt.title("Task 2: Linear Regression (5 Features)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
X_train_40 = np.column_stack((
    np.ones(X_train.shape[0]),
    X_train
))

X_test_40 = np.column_stack((
    np.ones(X_test.shape[0]),
    X_test
))

print("Training matrix:", X_train_40.shape)
print("Testing matrix:", X_test_40.shape)
```

```python
A_40 = X_train_40.T @ X_train_40

B_40 = X_train_40.T @ y_train

theta_40 = np.linalg.solve(A_40, B_40)

print("Estimated parameters:")
print(theta_40)

print("\nNumber of parameters:", len(theta_40))
```

```python
y_train_pred_40 = X_train_40 @ theta_40

y_test_pred_40 = X_test_40 @ theta_40

print("First five training predictions:")
print(y_train_pred_40[:5])

print("\nFirst five testing predictions:")
print(y_test_pred_40[:5])
```

```python
train_mse_40 = calculate_mse(
    y_train,
    y_train_pred_40
)

test_mse_40 = calculate_mse(
    y_test,
    y_test_pred_40
)

print("Training MSE:", train_mse_40)
print("Testing MSE:", test_mse_40)
```

```python
theta_norm_40 = np.linalg.norm(theta_40)

print("Euclidean norm:", theta_norm_40)
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    y_test,
    y_test_pred_40,
    label="Predictions"
)

minimum = min(y_test.min(), y_test_pred_40.min())
maximum = max(y_test.max(), y_test_pred_40.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    'r--',
    label="Perfect prediction"
)

plt.xlabel("Actual values")
plt.ylabel("Predicted values")

plt.title("Task 3: Linear Regression (40 Features)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
I = np.eye(X_train_40.shape[1])

n_train = X_train_40.shape[0]

print("Identity matrix shape:", I.shape)
print("Number of training observations:", n_train)
```

```python
lambda_values = [0.01, 0.1, 10]

print("Lambda values:", lambda_values)
```

```python
# Store experimental results
ridge_results = []

# Store trained parameter vectors
ridge_parameters = {}

# Calculate common components
A_base = X_train_40.T @ X_train_40

B = X_train_40.T @ y_train

# Train Ridge Regression for each lambda
for lam in lambda_values:

    # Regularised normal equation
    A_ridge = A_base + n_train * lam * I

    # Solve for parameters
    theta_ridge = np.linalg.solve(
        A_ridge,
        B
    )

    # Generate training predictions
    train_pred = X_train_40 @ theta_ridge

    # Generate testing predictions
    test_pred = X_test_40 @ theta_ridge

    # Calculate training MSE
    train_mse = calculate_mse(
        y_train,
        train_pred
    )

    # Calculate testing MSE
    test_mse = calculate_mse(
        y_test,
        test_pred
    )

    # Calculate Euclidean norm
    theta_norm = np.linalg.norm(theta_ridge)

    # Store results
    ridge_parameters[lam] = theta_ridge

    ridge_results.append([
        lam,
        train_mse,
        test_mse,
        theta_norm
    ])

    print("\nLambda:", lam)
    print("Training MSE:", train_mse)
    print("Testing MSE:", test_mse)
    print("Parameter norm:", theta_norm)
```

```python
# Create identity matrix
I = np.eye(X_train_40.shape[1])

# Number of training observations
n_train = X_train_40.shape[0]

# Lambda values
lambda_values = [0.01, 0.1, 10]

# Initialize result storage
ridge_results = []
ridge_parameters = {}

# Calculate common components
A_base = X_train_40.T @ X_train_40
B = X_train_40.T @ y_train

# Ridge Regression
for lam in lambda_values:

    A_ridge = A_base + n_train * lam * I

    theta_ridge = np.linalg.solve(A_ridge, B)

    train_pred = X_train_40 @ theta_ridge
    test_pred = X_test_40 @ theta_ridge

    train_mse = calculate_mse(y_train, train_pred)
    test_mse = calculate_mse(y_test, test_pred)

    theta_norm = np.linalg.norm(theta_ridge)

    ridge_parameters[lam] = theta_ridge

    ridge_results.append([
        lam,
        train_mse,
        test_mse,
        theta_norm
    ])

print("Ridge Regression completed successfully!")

print("\n", ridge_results)
```

```python
lambdas = [row[0] for row in ridge_results]
training_mse = [row[1] for row in ridge_results]
testing_mse = [row[2] for row in ridge_results]

plt.figure(figsize=(8, 5))

plt.semilogx(
    lambdas,
    training_mse,
    marker='o',
    linewidth=2,
    label='Training MSE'
)

plt.semilogx(
    lambdas,
    testing_mse,
    marker='s',
    linewidth=2,
    label='Testing MSE'
)

plt.xlabel("Lambda (Regularisation Strength)")
plt.ylabel("Mean Squared Error")

plt.title("Training and Testing MSE vs Lambda")

plt.xticks(lambdas, ['0.01', '0.1', '10'])
plt.legend()
plt.grid(True)

plt.show()
```

```python
theta_norms = [row[3] for row in ridge_results]

plt.figure(figsize=(8, 5))

plt.semilogx(
    lambdas,
    theta_norms,
    marker='o',
    linewidth=2,
    color='purple'
)

plt.xlabel("Lambda (Regularisation Strength)")
plt.ylabel("Euclidean Norm of Parameters")

plt.title("Parameter Norm vs Lambda")

plt.xticks(lambdas, ['0.01', '0.1', '10'])
plt.grid(True)

plt.show()
```
